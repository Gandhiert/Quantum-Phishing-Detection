## 3. Data Preprocessing & Quantum Downsampling
Classical hardware faces exponential RAM constraints when simulating quantum circuits. To guarantee computational efficiency on a classical simulator, the following data pipeline is executed:
* **Downsampling:** Reducing the dataset size to an optimized sample of $N=100$ rows.
* **Feature Selection:** Restricting the input to 2 core variables (`having_IP` and `URL_Length`) to minimize the required qubit count.
* **Normalization:** Applying `MinMaxScaler` to map the binary features into the range $[0, 1]$, making the data compatible with quantum gate rotation angles.


```python
import numpy as np
import pandas as pd
import os
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
from qiskit_aer import AerSimulator
from qiskit.circuit.library import ZZFeatureMap
from qiskit_machine_learning.algorithms import QSVC

# Otomatis nyari file csv/parquet di folder input Kaggle lu
path_data = ""
for dirname, _, filenames in os.walk('/kaggle/input'):
    for filename in filenames:
        if filename.endswith('.parquet') or filename.endswith('.csv'):
            path_data = os.path.join(dirname, filename)

df = pd.read_csv(path_data) if path_data.endswith('.csv') else pd.read_parquet(path_data)

# POTONG DATA: Ambil 100 baris acak biar laptop/server gak freeze
df_small = df.sample(n=100, random_state=42)

# Ambil 2 fitur biner utama: 'having_IP' dan 'URL_Length'
X_features = df_small[['having_IP', 'URL_Length']].values
y = df_small.iloc[:, -1].values

scaler = MinMaxScaler(feature_range=(0, 1))
X_scaled = scaler.fit_transform(X_features)

X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)
print("Data udah rapi dan siap diproses kuantum!")
```

## 4. Quantum Feature Map Construction & QSVC Training
This section implements the core components of the **Quantum Support Vector Machine (QSVM)** architecture:
* **`ZZFeatureMap`:** A second-order perturbative quantum feature map that encodes classical variables into a 2-qubit quantum state using linear entanglement.
* **`FidelityQuantumKernel`:** Maps the data into a high-dimensional *Hilbert Space* to compute the quantum kernel matrix based on state fidelities.
* **`QSVC`:** Initializes the Quantum Support Vector Classifier to optimize the decision hyperplane separating legitimate websites from phishing threats.


```python
from qiskit_machine_learning.kernels import FidelityQuantumKernel

# 1. Bikin sirkuit pemetaan kuantum (2 qubit karena kita pake 2 fitur data)
feature_map = ZZFeatureMap(feature_dimension=2, reps=1, entanglement='linear')

# 2. Bikin Quantum Kernel secara eksplisit (Ini inti komputasi kuantumnya)
q_kernel = FidelityQuantumKernel(feature_map=feature_map)

# 3. Inisialisasi model QSVC pake argumen 'quantum_kernel' yang sah
qsvc = QSVC(quantum_kernel=q_kernel)

# 4. Mulai proses training kuantum
print("Sabar, lagi ditraining kuantum pake FidelityQuantumKernel...")
qsvc.fit(X_train, y_train)
print("Training kelar tanpa error, bre!")
```

## 5. Model Evaluation & Accuracy Metrics
This final section evaluates the trained quantum model on both the training and testing subsets to extract accuracy metrics, providing empirical proof-of-concept performance data for the cybersecurity task.


```python
# Menghitung skor akurasi model kuantum kelompok lu
train_acc = qsvc.score(X_train, y_train)
test_acc = qsvc.score(X_test, y_test)

print(f"Akurasi Data Training: {train_acc * 100:.2f}%")
print(f"Akurasi Deteksi Phishing Baru (Testing): {test_acc * 100:.2f}%")
```
